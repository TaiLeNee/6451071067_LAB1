Câu 2: Xóa ký tự thừa ở Description, xóa các dòng không có InvoiceNo và chuyển InvoiceNo sang kiểu chuỗi.

```python
print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 2")

du_lieu_cau2 = df.copy()

du_lieu_cau2["Description"] = du_lieu_cau2["Description"].str.strip()

so_dong_truoc = len(du_lieu_cau2)
du_lieu_cau2 = du_lieu_cau2.dropna(subset=["InvoiceNo"]).copy()

du_lieu_cau2["InvoiceNo"] = du_lieu_cau2["InvoiceNo"].astype("string")

print("\nSố dòng trước xử lý:", so_dong_truoc)
print("Số dòng đã xóa:", so_dong_truoc - len(du_lieu_cau2))
print("Số dòng còn lại:", len(du_lieu_cau2))
print("Kiểu dữ liệu của InvoiceNo:", du_lieu_cau2["InvoiceNo"].dtype)

print("\nMười dòng dữ liệu sau khi xử lý")
print(du_lieu_cau2[["InvoiceNo", "Description", "Quantity"]].head(10).to_string(index=False))
```

- Nhận xét: Đã chuẩn hóa tên mặt hàng, loại bỏ các hóa đơn thiếu mã và chuyển InvoiceNo sang dạng chuỗi.

Câu 3: Loại bỏ hóa đơn tín dụng có chứa ký tự C trong InvoiceNo.

```python
print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 3")

hoa_don_tin_dung = du_lieu_cau2["InvoiceNo"].str.contains("C", na=False)

print("\nMột số hóa đơn tín dụng trước khi loại bỏ")
print(du_lieu_cau2.loc[hoa_don_tin_dung, ["InvoiceNo", "Description", "Quantity"]].head(5).to_string(index=False))

du_lieu_cau3 = du_lieu_cau2.loc[~hoa_don_tin_dung].copy()

print("\nSố dòng thuộc hóa đơn tín dụng:", hoa_don_tin_dung.sum())
print("Số dòng còn lại:", len(du_lieu_cau3))
print("Số dòng có InvoiceNo chứa C sau xử lý:", du_lieu_cau3["InvoiceNo"].str.contains("C", na=False).sum())
```

- Nhận xét: Những hóa đơn có mã chứa C đã được loại bỏ khỏi dữ liệu giao dịch dùng cho các bước tiếp theo.

Câu 4: Thống kê số dòng dữ liệu theo từng quốc gia.

```python
print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 4")

thong_ke_quoc_gia = du_lieu_cau3["Country"].value_counts(dropna=False)

print("\nThống kê số dòng dữ liệu theo từng quốc gia")
print(thong_ke_quoc_gia.to_string())

print("\nTổng số dòng thống kê:", thong_ke_quoc_gia.sum())
```

Câu 5: Lấy hóa đơn tại United Kingdom và gom nhóm Quantity theo InvoiceNo và Description.

```python
print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 5")

du_lieu_anh = du_lieu_cau3.loc[
    du_lieu_cau3["Country"] == "United Kingdom"
].copy()

du_lieu_gom_nhom = (
    du_lieu_anh
    .groupby(["InvoiceNo", "Description"], as_index=False)["Quantity"]
    .sum()
)

print("\nSố dòng dữ liệu từ United Kingdom:", len(du_lieu_anh))
print("Số cặp hóa đơn và mặt hàng sau khi gom nhóm:", len(du_lieu_gom_nhom))

print("\nMười dòng kết quả gom nhóm")
print(du_lieu_gom_nhom.head(10).to_string(index=False))
```

- Nhận xét: Dữ liệu được lọc theo United Kingdom và tổng hợp số lượng từng mặt hàng trên từng hóa đơn. Mỗi dòng kết quả ứng với một cặp hóa đơn và mặt hàng.

Câu 6: Chuyển dữ liệu sang dạng hot encoding, mỗi dòng là một hóa đơn.

```python
from scipy.sparse import coo_matrix

print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 6")

danh_sach_hoa_don = pd.Index(du_lieu_gom_nhom["InvoiceNo"].unique())
danh_sach_mat_hang = pd.Index(du_lieu_gom_nhom["Description"].unique())

vi_tri_hoa_don = pd.Categorical(
    du_lieu_gom_nhom["InvoiceNo"], categories=danh_sach_hoa_don
).codes
vi_tri_mat_hang = pd.Categorical(
    du_lieu_gom_nhom["Description"], categories=danh_sach_mat_hang
).codes

ma_tran_gio_hang = coo_matrix(
    (du_lieu_gom_nhom["Quantity"].to_numpy(), (vi_tri_hoa_don, vi_tri_mat_hang)),
    shape=(len(danh_sach_hoa_don), len(danh_sach_mat_hang))
).tocsr()

gio_hang = pd.DataFrame.sparse.from_spmatrix(
    ma_tran_gio_hang,
    index=danh_sach_hoa_don,
    columns=danh_sach_mat_hang
)
gio_hang.index.name = "InvoiceNo"

print("\nKích thước bảng giỏ hàng")
print("Số hóa đơn:", gio_hang.shape[0])
print("Số mặt hàng:", gio_hang.shape[1])

print("\nMẫu bảng hot encoding (5 mặt hàng đầu)")
mau_gio_hang = gio_hang.iloc[:5, :5].copy()
mau_gio_hang.columns = [ten[:15] for ten in mau_gio_hang.columns]
print(mau_gio_hang.to_string())
print("\nCác ô chứa tổng số lượng mua, chưa chuyển thành 0 hoặc 1")
```

Câu 7: Chuyển bảng hot encoding thành one-hot encoding (0 và 1).

```python
print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 7")

one_hot = gio_hang.gt(0)

print("\nKích thước dữ liệu one-hot encoding")
print("Số hóa đơn:", one_hot.shape[0])
print("Số mặt hàng:", one_hot.shape[1])
print("Kiểu dữ liệu đầu tiên:", one_hot.dtypes.iloc[0])

print("\nNăm dòng đầu của bảng one-hot (5 mặt hàng đầu)")
mau_one_hot = one_hot.iloc[:5, :5].astype(int)
mau_one_hot.columns = [ten[:15] for ten in mau_one_hot.columns]
print(mau_one_hot.to_string())
```

Câu 8: Xóa cột POSTAGE vì đây là tiền cước phí, không phải mặt hàng.

```python
print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 8")

print("\nSố mặt hàng trước khi xử lý:", one_hot.shape[1])

one_hot_san_pham = one_hot.drop(columns=["POSTAGE"], errors="ignore")

print("Số mặt hàng sau khi xử lý:", one_hot_san_pham.shape[1])
print("Còn cột POSTAGE:", "POSTAGE" in one_hot_san_pham.columns)
print("Số hóa đơn sau khi xử lý:", one_hot_san_pham.shape[0])
```

Câu 9: Tìm tập phổ biến bằng thuật toán Apriori với min_sup = 3%.

```python
from mlxtend.frequent_patterns import apriori

print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 9")

tap_pho_bien_apriori = apriori(
    one_hot_san_pham,
    min_support=0.03,
    use_colnames=True
)

tap_pho_bien_apriori = tap_pho_bien_apriori.sort_values(
    "support", ascending=False
).reset_index(drop=True)

print("\nSố tập phổ biến tìm được:", len(tap_pho_bien_apriori))
print("\nMười tập phổ biến có độ hỗ trợ cao nhất")
print(tap_pho_bien_apriori.head(10).to_string(index=False))
```

Câu 10: Tạo luật kết hợp có min_conf = 50% và in kết quả.

```python
from mlxtend.frequent_patterns import association_rules

print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 10")

luat_apriori = association_rules(
    tap_pho_bien_apriori,
    metric="confidence",
    min_threshold=0.5,
    num_itemsets=len(one_hot_san_pham)
)

cot_luat = ["antecedents", "consequents", "support", "confidence", "lift"]
luat_apriori = luat_apriori.sort_values(
    ["confidence", "support"], ascending=False
).reset_index(drop=True)

print("\nTổng số luật kết hợp đạt yêu cầu:", len(luat_apriori))
print("\nCác luật kết hợp tìm được")
print(luat_apriori[cot_luat].to_string(index=False))
```

Câu 11: Vẽ biểu đồ phân tán độ hỗ trợ và độ tin cậy của các luật kết hợp.

```python
import matplotlib.pyplot as plt

print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 11")

plt.figure(figsize=(9, 6))
plt.scatter(
    luat_apriori["support"],
    luat_apriori["confidence"],
    alpha=0.8,
    s=65
)
plt.xlabel("Độ hỗ trợ (support)")
plt.ylabel("Độ tin cậy (confidence)")
plt.title("Phân bố độ hỗ trợ và độ tin cậy của các luật kết hợp")
plt.grid(alpha=0.25)
plt.tight_layout()
plt.show()

print("\nSố luật được biểu diễn:", len(luat_apriori))
```

Câu 12: Dùng FP-Growth với min_sup = 3%, min_conf = 50% và so sánh với Apriori.

```python
from mlxtend.frequent_patterns import fpgrowth

print("Họ và tên: Lê Công Tài")
print("MSSV: 6451071067")
print("Lab 1, Phần 2, Bài 2, Câu 12")

tap_pho_bien_fp = fpgrowth(
    one_hot_san_pham,
    min_support=0.03,
    use_colnames=True
)

tap_pho_bien_fp = tap_pho_bien_fp.sort_values(
    "support", ascending=False
).reset_index(drop=True)

luat_fp = association_rules(
    tap_pho_bien_fp,
    metric="confidence",
    min_threshold=0.5,
    num_itemsets=len(one_hot_san_pham)
)
luat_fp = luat_fp.sort_values(
    ["confidence", "support"], ascending=False
).reset_index(drop=True)

print("\nSố tập phổ biến tìm bằng FP-Growth:", len(tap_pho_bien_fp))
print("Số luật kết hợp từ FP-Growth:", len(luat_fp))
print("\nMười tập phổ biến đầu tiên của FP-Growth")
print(tap_pho_bien_fp.head(10).to_string(index=False))
print("\nCác luật FP-Growth tìm được")
print(luat_fp[cot_luat].to_string(index=False))

ho_tro_apriori = dict(zip(
    tap_pho_bien_apriori["itemsets"], tap_pho_bien_apriori["support"]
))
ho_tro_fp = dict(zip(tap_pho_bien_fp["itemsets"], tap_pho_bien_fp["support"]))

cung_tap_pho_bien = ho_tro_apriori.keys() == ho_tro_fp.keys()
cung_do_ho_tro = cung_tap_pho_bien and all(
    abs(ho_tro_apriori[items] - ho_tro_fp[items]) < 1e-10
    for items in ho_tro_apriori
)

ma_luat_apriori = {
    (frozenset(a), frozenset(b))
    for a, b in zip(luat_apriori["antecedents"], luat_apriori["consequents"])
}
ma_luat_fp = {
    (frozenset(a), frozenset(b))
    for a, b in zip(luat_fp["antecedents"], luat_fp["consequents"])
}

print("\nSo sánh Apriori và FP-Growth")
print("Số tập phổ biến của Apriori:", len(tap_pho_bien_apriori))
print("Số tập phổ biến của FP-Growth:", len(tap_pho_bien_fp))
print("Số luật của Apriori:", len(luat_apriori))
print("Số luật của FP-Growth:", len(luat_fp))
print("Tập phổ biến và độ hỗ trợ trùng khớp:", cung_do_ho_tro)
print("Các luật kết hợp trùng khớp:", ma_luat_apriori == ma_luat_fp)
```
